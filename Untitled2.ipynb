```python
# ===================================================================
# 📘 Step 3: EXPLORATORY DATA ANALYSIS (EDA)
# ===================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder, MinMaxScaler
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.metrics import accuracy_score, mean_squared_error, r2_score

# -------------------------------------------------------------------
# Load dataset
data = pd.read_csv("student_data.csv")   # Replace with your dataset filename
print("✅ Dataset Loaded Successfully\n")

# -------------------------------------------------------------------
# Basic structure of the dataset
print("📊 Dataset Info:")
print(data.info())

print("\n📈 Summary Statistics:")
print(data.describe())

print("\n🔍 First 5 rows of the dataset:")
print(data.head())

# -------------------------------------------------------------------
# Check for missing values and duplicates
print("\n🧩 Missing Values per Column:")
print(data.isnull().sum())

print("\n📋 Number of Duplicates:", data.duplicated().sum())

# -------------------------------------------------------------------
# Visualize distributions and relationships
plt.figure(figsize=(6,4))
sns.heatmap(data.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()

# Histogram for all numeric columns
data.hist(bins=30, figsize=(12,8))
plt.suptitle("Feature Distributions", fontsize=14)
plt.show()

# Pairplot to visualize relationships (only if dataset is small)
# sns.pairplot(data)
# plt.show()

# ===================================================================
# 📘 Step 4: DATA PREPROCESSING
# ===================================================================

# -------------------------------------------------------------------
# Handle Missing Values (fill with mean for numeric, mode for categorical)
for col in data.columns:
    if data[col].dtype == 'object':
        data[col].fillna(data[col].mode()[0], inplace=True)
    else:
        data[col].fillna(data[col].mean(), inplace=True)

# -------------------------------------------------------------------
# Convert categorical columns to numeric using Label Encoding
label_enc = LabelEncoder()
for col in data.select_dtypes(include=['object']).columns:
    data[col] = label_enc.fit_transform(data[col])

# -------------------------------------------------------------------
# Feature Scaling / Normalization
scaler = StandardScaler()
scaled_features = scaler.fit_transform(data.select_dtypes(include=np.number))
data_scaled = pd.DataFrame(scaled_features, columns=data.select_dtypes(include=np.number).columns)

print("\n✅ Data Preprocessing Completed")

# ===================================================================
# 📘 Step 5: FEATURE ENGINEERING / EXTRACTION
# ===================================================================

# Example: extract features from a 'Date' column if present
if 'Date' in data.columns:
    data['Date'] = pd.to_datetime(data['Date'])
    data['Year'] = data['Date'].dt.year
    data['Month'] = data['Date'].dt.month
    data['Day'] = data['Date'].dt.day
    data.drop(columns=['Date'], inplace=True)

# Example: Feature Selection (keep top correlated features)
corr = data.corr(numeric_only=True)
target = corr.columns[-1]  # assume last column is target
important_features = corr[target].sort_values(ascending=False)
print("\n🔥 Important Features for Target:\n", important_features)

# ===================================================================
# 📘 Step 6: MODEL SELECTION
# ===================================================================

# -------------------------------------------------------------------
# Train-Test Split (70% train, 30% test)
X = data.drop(columns=[data.columns[-1]])  # features
y = data[data.columns[-1]]                 # target variable

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# -------------------------------------------------------------------
# Detect whether it's regression or classification
if y.dtype == 'float64' or len(y.unique()) > 10:
    problem_type = "regression"
else:
    problem_type = "classification"

print(f"\n🧠 Problem Type Detected: {problem_type.upper()}")

# -------------------------------------------------------------------
# Choose and train models accordingly
if problem_type == "regression":
    models = {
        "Linear Regression": LinearRegression(),
        "Random Forest Regressor": RandomForestRegressor()
    }

    for name, model in models.items():
        model.fit(X_train, y_train)
        preds = model.predict(X_test)
        mse = mean_squared_error(y_test, preds)
        r2 = r2_score(y_test, preds)
        print(f"\n{name} Results:")
        print(f"Mean Squared Error: {mse:.2f}")
        print(f"R² Score: {r2:.2f}")

else:
    models = {
        "Logistic Regression": LogisticRegression(max_iter=1000),
        "Random Forest Classifier": RandomForestClassifier()
    }

    for name, model in models.items():
        model.fit(X_train, y_train)
        preds = model.predict(X_test)
        acc = accuracy_score(y_test, preds)
        print(f"\n{name} Accuracy: {acc*100:.2f}%")

print("\n✅ Model Training & Evaluation Completed Successfully!")
```

```python
# ===================================================================
# 📘 Step 7: MODEL TRAINING (Fixed with Encoding)
# ===================================================================

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# -------------------------------------------------------------------
# Load dataset
data = pd.read_csv("student_data.csv")
print("✅ Dataset Loaded Successfully\n")

# -------------------------------------------------------------------
# Check for non-numeric columns
print("🧩 Checking Data Types:")
print(data.dtypes)

# -------------------------------------------------------------------
# Encode categorical (string) columns to numeric
label_enc = LabelEncoder()
for col in data.select_dtypes(include=['object']).columns:
    data[col] = label_enc.fit_transform(data[col])

print("\n✅ All categorical columns converted to numeric form!")
print(data.head())

# -------------------------------------------------------------------
# Separate features (X) and target variable (y)
# ⚠️ Replace 'target' with your actual column name
X = data.drop("failures", axis=1)
y = data["failures"]

# -------------------------------------------------------------------
# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

print("\n✅ Data split completed successfully!")

# -------------------------------------------------------------------
# Initialize and train the model
model = DecisionTreeClassifier(random_state=42)
print("\n🌳 Decision Tree Classifier Model Initialized")

model.fit(X_train, y_train)
print("\n🚀 Model Training Completed Successfully!")

# -------------------------------------------------------------------
# Model Prediction
y_pred = model.predict(X_test)

# -------------------------------------------------------------------
# Evaluate Model
print("\n🎯 MODEL EVALUATION RESULTS")
print("-----------------------------")
print(f"✅ Accuracy: {accuracy_score(y_test, y_pred)*100:.2f}%")

print("\n📋 Classification Report:")
print(classification_report(y_test, y_pred))

print("\n🧩 Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
# ===================================================================
# 📘 STEP 8: MODEL EVALUATION
# ===================================================================

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    mean_squared_error, r2_score
)
import numpy as np

# -------------------------------------------------------------------
# Load dataset
data = pd.read_csv("student_data.csv")   # Replace with your actual dataset
print("✅ Dataset Loaded Successfully\n")

# -------------------------------------------------------------------
# Encode categorical (string) columns to numeric
label_enc = LabelEncoder()
for col in data.select_dtypes(include=['object']).columns:
    data[col] = label_enc.fit_transform(data[col])

# -------------------------------------------------------------------
# Separate features (X) and target variable (y)
# ⚠️ Replace 'target' with your actual target column name
X = data.drop("failures", axis=1)
y = data["failures"]

# -------------------------------------------------------------------
# Split data into train and test sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# -------------------------------------------------------------------
# Automatically detect problem type (Classification or Regression)
if y.nunique() <= 10 and y.dtypes != 'float64':
    problem_type = "classification"
    model = DecisionTreeClassifier(random_state=42)
else:
    problem_type = "regression"
    model = DecisionTreeRegressor(random_state=42)

print(f"🧠 Problem Type Detected: {problem_type.upper()}")
print("🌳 Model Initialized")

# -------------------------------------------------------------------
# Train the model
model.fit(X_train, y_train)
print("\n🚀 Model Training Completed Successfully!")

# -------------------------------------------------------------------
# Predict using the trained model
y_pred = model.predict(X_test)

# -------------------------------------------------------------------
# EVALUATE MODEL PERFORMANCE
print("\n📊 ===============================================")
print("📘 STEP 8: MODEL EVALUATION RESULTS")
print("===============================================\n")

if problem_type == "classification":
    # Compute classification metrics
    acc = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred, average='weighted')
    rec = recall_score(y_test, y_pred, average='weighted')
    f1 = f1_score(y_test, y_pred, average='weighted')

    print("🎯 CLASSIFICATION METRICS")
    print("---------------------------")
    print(f"✅ Accuracy Score  : {acc*100:.2f}%")
    print(f"✅ Precision Score : {prec*100:.2f}%")
    print(f"✅ Recall Score    : {rec*100:.2f}%")
    print(f"✅ F1-Score        : {f1*100:.2f}%")

else:
    # Compute regression metrics
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)

    print("📈 REGRESSION METRICS")
    print("---------------------------")
    print(f"✅ Mean Squared Error (MSE): {mse:.2f}")
    print(f"✅ Root Mean Squared Error (RMSE): {rmse:.2f}")
    print(f"✅ R² Score: {r2:.2f}")

print("\n✅ Model Evaluation Completed Successfully!")
print("===============================================\n")
```

```python
# ===================================================================
# 📘 STEP 9: MODEL TUNING (OPTIMIZATION)
# ===================================================================

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, GridSearchCV, cross_val_score
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    mean_squared_error, r2_score
)

# -------------------------------------------------------------------
# Optional: Import XGBoost (make sure xgboost is installed)
try:
    from xgboost import XGBClassifier, XGBRegressor
    xgb_available = True
except ImportError:
    xgb_available = False
    print("⚠️ XGBoost not installed — skipping XGBoost section.")

# -------------------------------------------------------------------
# Load dataset
data = pd.read_csv("student_data.csv")   # Replace with your dataset
print("✅ Dataset Loaded Successfully\n")

# -------------------------------------------------------------------
# Encode categorical columns
label_enc = LabelEncoder()
for col in data.select_dtypes(include=['object']).columns:
    data[col] = label_enc.fit_transform(data[col])

# -------------------------------------------------------------------
# Separate features and target
# ⚠️ Replace 'target' with your actual target column name
X = data.drop("failures", axis=1)
y = data["failures"]

# -------------------------------------------------------------------
# Split into train and test sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# -------------------------------------------------------------------
# Detect problem type automatically
if y.nunique() <= 10 and y.dtypes != 'float64':
    problem_type = "classification"
else:
    problem_type = "regression"

print(f"🧠 Problem Type Detected: {problem_type.upper()}")

# ===================================================================
# ⚙️ HYPERPARAMETER TUNING using GridSearchCV
# ===================================================================

print("\n🔍 Starting Hyperparameter Tuning using GridSearchCV...")
if problem_type == "classification":
    base_model = DecisionTreeClassifier(random_state=42)
    param_grid = {
        'criterion': ['gini', 'entropy'],
        'max_depth': [3, 5, 7, 10],
        'min_samples_split': [2, 5, 10]
    }
else:
    base_model = DecisionTreeRegressor(random_state=42)
    param_grid = {
        'criterion': ['squared_error', 'friedman_mse'],
        'max_depth': [3, 5, 7, 10],
        'min_samples_split': [2, 5, 10]
    }

grid_search = GridSearchCV(
    estimator=base_model,
    param_grid=param_grid,
    cv=5,
    scoring='accuracy' if problem_type == "classification" else 'r2',
    n_jobs=-1
)

grid_search.fit(X_train, y_train)
best_model = grid_search.best_estimator_

print("\n🏆 Best Parameters Found:")
print(grid_search.best_params_)
print("✅ Model Tuned Successfully!")

# ===================================================================
# 🔁 CROSS-VALIDATION (5-Fold)
# ===================================================================
print("\n📊 Performing 5-Fold Cross Validation...")
scores = cross_val_score(
    best_model, X_train, y_train, cv=5,
    scoring='accuracy' if problem_type == "classification" else 'r2'
)

print(f"✅ Cross-Validation Scores: {scores}")
print(f"📈 Average CV Score: {np.mean(scores):.4f}")

# ===================================================================
# 🌲 ENSEMBLE MODELS (Random Forest + XGBoost)
# ===================================================================
print("\n🌳 Training Ensemble Models for Comparison...")

if problem_type == "classification":
    rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
    rf_model.fit(X_train, y_train)
    rf_pred = rf_model.predict(X_test)
    rf_acc = accuracy_score(y_test, rf_pred)

    print(f"✅ Random Forest Accuracy: {rf_acc*100:.2f}%")

    if xgb_available:
        xgb_model = XGBClassifier(random_state=42, eval_metric='logloss')
        xgb_model.fit(X_train, y_train)
        xgb_pred = xgb_model.predict(X_test)
        xgb_acc = accuracy_score(y_test, xgb_pred)
        print(f"✅ XGBoost Accuracy: {xgb_acc*100:.2f}%")

else:
    rf_model = RandomForestRegressor(n_estimators=100, random_state=42)
    rf_model.fit(X_train, y_train)
    rf_pred = rf_model.predict(X_test)
    rf_r2 = r2_score(y_test, rf_pred)
    print(f"✅ Random Forest R² Score: {rf_r2:.2f}")

    if xgb_available:
        xgb_model = XGBRegressor(random_state=42)
        xgb_model.fit(X_train, y_train)
        xgb_pred = xgb_model.predict(X_test)
        xgb_r2 = r2_score(y_test, xgb_pred)
        print(f"✅ XGBoost R² Score: {xgb_r2:.2f}")

# -------------------------------------------------------------------
print("\n🎯 MODEL TUNING & OPTIMIZATION COMPLETED SUCCESSFULLY!")
print("===========================================================\n")
```

```python
# ==============================================================
# 🧠 STEP 10: MODEL DEPLOYMENT / FINAL PRESENTATION
# ==============================================================

# Import necessary libraries
import pickle
from flask import Flask, request, jsonify
import numpy as np

# --------------------------------------------------------------
# 🎯 1. SAVE TRAINED MODEL
# --------------------------------------------------------------
# Let's assume 'best_model' from previous tuning step is your final model.
# You can replace 'best_model' with your trained model variable (e.g., model, rf, xgb, etc.)

# Save model using pickle
with open('final_model.pkl', 'wb') as file:
    pickle.dump(best_model, file)

print("✅ Model saved successfully as 'final_model.pkl'!")

# --------------------------------------------------------------
# ⚙️ 2. CREATE SIMPLE FLASK APP FOR DEPLOYMENT
# --------------------------------------------------------------
app = Flask(__name__)

# Load model
with open('final_model.pkl', 'rb') as file:
    loaded_model = pickle.load(file)

@app.route('/')
def home():
    return "🚀 ML Model Deployment Successful! Use /predict endpoint for predictions."

@app.route('/predict', methods=['POST'])
def predict():
    # Get input data from user
    data = request.get_json(force=True)

    # Example: expecting {"features": [5.1, 3.5, 1.4, 0.2]} for Iris dataset
    features = np.array(data['features']).reshape(1, -1)

    # Predict using loaded model
    prediction = loaded_model.predict(features)

    # Return result
    return jsonify({"prediction": prediction.tolist()})

if __name__ == '__main__':
    print("\n🌐 Starting Flask app... Visit http://127.0.0.1:5000")
    app.run(debug=True)
```
